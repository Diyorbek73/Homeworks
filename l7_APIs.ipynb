API (Application Programming Interface — программный интерфейс приложения) — это набор правил, по которым одна программа может взаимодействовать с другой.

| API Type      | Who Can Use It        | Example Use                        | Access Type    |
| ------------- | --------------------- | ---------------------------------- | -------------- |
| Open API      | Anyone                | Get weather or music info          | Public         |
| Internal API  | Only company staff    | Connect HR with payroll system     | Private        |
| Composite API | Developers (anywhere) | One API call for multiple services | Mixed          |
| Partner API   | Business partners     | Amazon sellers managing inventory  | Restricted/VIP |


**Sample API's**

REST (Representational State Transfer — передача состояния представления) — это архитектурный стиль, определяющий, как писать серверный код таким образом, чтобы системы могли легко обмениваться данными, а само приложение — масштабироваться.

| HTTP Method | Purpose (What it Does)                           | Example Request                   | Example Use Case                            |
| ----------- | ------------------------------------------------ | --------------------------------- | ------------------------------------------- |
| **GET**     | Retrieve (read) data from the server             | `GET /users`                      | Get a list of all users                     |
| **POST**    | Send (create) new data to the server             | `POST /users` with JSON body      | Create a new user account                   |
| **PUT**     | Update existing data (replace fully)             | `PUT /users/123` with JSON body   | Replace all info for user 123               |
| **PATCH**   | Update part of existing data (modify partially)  | `PATCH /users/123` with JSON body | Change only the email of user 123           |
| **DELETE**  | Remove data from the server                      | `DELETE /users/123`               | Delete user 123                             |
| **HEAD**    | Get only the response headers (no body)          | `HEAD /users`                     | Check if data exists without downloading it |
| **OPTIONS** | Find out what methods are allowed for a resource | `OPTIONS /users`                  | Ask the server: “What can I do here?”       |
