Quyida talabaning baholariga qarab “o‘tdi” yoki “o‘tmadi” natijasini hisoblaydigan dastur berilgan. Ushbu kodni `is_passed` deb nomlangan funktsiyaga o‘rab qo‘ying. Funktsiya baholar ro‘yxatini kirish sifatida qabul qilsin va mantiqiy qiymat (True – o‘tdi, False – o‘tmadi) qaytarsin.

Keyingi kataklarda `is_passed` funktsiyasini chaqirishlar to‘g‘ri ishlashini tekshiring.

[5, 4, 2, 5, 1] -> [True, False, False, True, False]

```python
def is_passed(baholar) :
    res = []
    for i in baholar:
        if i == 5 :
            res.append(True) 
        else : 
            res.append(False)
    return res
```

```python
is_passed([5, 5, 4, 4])
```

```python
is_passed([5, 5, 4, 4, 3, 2])
```

```python
is_passed([5, 5, 5, 5, 3, 2])
```
